**LIST OPERATIONS IN PYTHON**

```python
#Concatenation
L1 = [2,4,6,8]
L2 = [1,3,5,7]
print(L1 + L2)
#Repetition
print(L1 * 4)
#indexing
print(L1[0])
#Slicing
print(L1[0:2])
#membership operator
print(4 in L1)
print(5 not in L2)
#Iteration
for i in L1:
  print(i,end = ' ')
```

**BUILT IN FUNCTIONS**

```python
L1 = [2,4,6,8]
print(len(L1))
print(sum(L1))
print(min(L1))
print(max(L1))
print(list(reversed(L1)))
print(sorted(L1))
```

**LIST METHODS(BUILT IN FUCTIONS)**

```python
L1 = [1,2,3]
L2 = [12,13,14]
L1.append(8)
print(L1)
L1.extend(L2)
print(L1)
L1.insert(0,0)
print(L1)
L1.pop(3)#index value
print(L1)
L1.remove(8)
print(L1)
L1.sort()
print(L1)
print(L1.count(1))
L1.reverse()
print(L1)
print(L1.index(14))
```

```python
#program that print sum of list elements without using sum()
L1 = [11,12,13,14,15,16,17,18]
sum = 0
for i in L1:
  sum = sum + i
print(sum)
```

```python
#print the list elements in odd index
L2 = [1,2,3,4,5,6,7,8]
for i in range(1,len(L2),2):
  print(L2[i],end = ' ')
```

```python
#print the even elements
L2 = [1,2,6,4,5]
for i in L2:
  if(i%2 == 0):
    print(i,end = ' ')
```

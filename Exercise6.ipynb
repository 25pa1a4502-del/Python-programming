```python
a = (1,2,'Abhi')
b = (3,4,'Ram')
print(a + b)
print(a * 2)
for i in a:
  print(i)
print(2 in a)
```

```python
a = (1,1,2,3,3,3,'Abhi')
print(a.count(3))
print(a.index('Abhi'))
```

```python
a = ('n1','n2','n3')
b = (1,2,3)
print(tuple(zip(a,b)))
```

```python
#program that reverse a string using functions
def reverse(a):
  print(a[::-1])
a = input('Enter a string ')
reverse(a)
```

```python
#Example code on default arguments
def print_msg(name,branch = 'AI&DS'):
  print('The student name is',name,'and the branch is',branch)
print_msg('abhi')
print_msg('ram')
```

```python
#Example code on keyword arguments
def subtraction(a,b,c):
  return a-b-c
res = subtraction(7,2,3)
print(res)
```

```python
#functions that returns multiple values
def multiple_values(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_len,res_avg = multiple_values(1,2,3,4,5)
print(f'The sum of numbers is {res_sum}, the count is {res_len}, the avg is {res_avg}')
```

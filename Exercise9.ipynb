```python
file = open('abhi.txt','w')
file.write('Hello World')
print('File created')
file.close()
```

```python
with open('abhi.txt','r') as f:
  data = f.read()
  print(data)
```

```python
with open('abhi.txt','a') as f:
  f.writelines(['\napple','\nname','\nend'])
```

```python
with open('abhi.txt','r') as f:
  print(f.readline())
```

```python
with open('abhi.txt','r') as f:
  print(f.readlines())
```

```python
with open('abhi.txt','r') as f:
  print(f.readlines())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
#copying content of one file to another
with open('abhi.txt','r') as f:
  data = f.read()
  print('Reading done')
with open('ram.txt','w') as f1:
  f1.write(data)
  print('File copied')
```

```python
#program that prints each line in a file in reverse order
with open('abhi.txt','r') as f:
  data = f.readlines()
  for i in data:
    print(i[::-1])
```

```python
#write a program to sort words in a file
#file should have only lower case words
#must be lowered
with open('abhi.txt','r') as f:
  data = f.read().split()
  data_lower = [i.lower() for i in data]
  sort_data = sorted(data_lower)
with open('abhi2.txt','w') as f1:
  for i in sort_data:
    f1.write(i+'\n')
  print('done')
```

Program that finds maximum among three numbers

```python
a = int(input('Enter the first number'))
b = int(input('Enter the second number'))
c = int(input('Enter the third number'))
if(a>b and a>c):
  print(a ,'is maximum among three numbers')
elif(b>c):
  print(b ,'is maximum among three numbers')
else:
  print(c ,'is maximum among three numbers')
```

Program to find given year is leap year or not

```python
year = int(input('Enter a year'))
if(year%4==0 and year%100!=0) or (year%400==0):
  print(f'{year} is a leap year')
else:
  print(f'{year} is not a leap year')
```

Program on match case

```python
day = int(input('Enter'))
match day:
  case 1:
    print('Sunday')
  case 2:
    print('Monday')
  case 3:
    print('Tuesday')
  case 4:
    print('Wednesday')
  case 5:
    print('Thursday')
  case 6:
    print('Friday')
  case 7:
    print('Saturday')
```

Adding and multiplying two complex numbers

```python
i = complex(input('Enter first number'))
j = complex(input('Enter second number'))
sum = i + j
mul = i * j
print(f'sum is {sum}')
print(f'mul is {mul}')
```

Swap two variables without using temporary variable

```python
x = int(input('Enter number '))
y = int(input('Enter another number'))
print(x,y)
x,y = y,x
print(x,y)
```

Print 1 to 10 numbers

```python
a = 1
while a<11:
  print(a)
  a+=1
```

Print n to 1

```python
b = int(input('Enter the number'))
while b>0:
  print(b,end=' ')
  b-=1
```

Print 1 to n even numbers only

```python
c = int(input('Enter the number'))
while c>=1:
  if c%2==0:
    print(c,end=' ')
  c-=1
```

For loop

```python
#print 1 to 10
for i in range(1,11):
  print(i,end=' ')
```

```python
#print n to 1
b = int(input('Enter a number'))
for i in range(b,0,-1):
  print(i,end=' ')
```

```python
#sum of digits of a given number
e = int(input('Enter a number'))
sum = 0
rem = 0
while e!=0:
  rem = e%10
  sum = sum + rem
  e = e//10
print(sum)
```

```python
#reverse
f = int(input('Enter a number'))
sum = 0
rem = 0
while f!=0:
  rem = f%10
  sum = sum*10 + rem
  f = f//10
print(sum)
```

```python
#prime or not
g = int(input())
for i in range(2,g+1):
  if g%i==0:
    print('not a prime number')
    break
else:
    print('prime number')
```

```python
#print prime numbers between two numbers
h = int(input())
k = int(input())
for i in range(h,k+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=' ')
```

**DICTIONARIES**

```python
#Creating dictionaries
d = {'rno':502,'Name':'Abhiram','Life':'Single'}
print(d)
print(d['Name'])
d['Name'] = 'AK47'
print(d)
```

```python
students = {'stu1':['Madhu',18],'stu2':['Abhi',19],'stu3':['Ram',19]}
print(students)
print(students['stu2'])
print(students.keys())
print(students.values())
```

**Built in functions**

```python
a = {'Name':'Abhi','Marks':99,'Life':'Single'}
print(len(a))
print(sorted(a))
print(max(a))
print(min(a))
print(list(a))
```

```python
#write a program to check if a given key exists in a dictionary or not
a = {'Name':'Abhi','rno':4502}
b = input()
if b in a:
  print('Yes')
else:
  print('No')
```

```python
#write a program to add new key-value pair to an existing dictionary
a = {'Name':'Abhi','rno':4502}
a['grade'] = 8
print(a)
a.update({'Height':6,'Village':'Penugonda'})
print(a)
```

```python
a = {1:10,2:20,3:30}
print(sum(a.values()))
```

```python
#write a program that prints frequency of each element in a list in following
#format [1,2,3,4,1,5,6,1,6,7]            {1:3,2:1,3:1,4:1,5:1,6:2,7:1}
a = [int(i) for i in input().split()]
d = {}
for i in a:
  d[i] = a.count(i)
print(d)
```
